### Question 3

#### 3a

```python
# t = x/(y - 1)
t = sp.Symbol('t', real=True)
x = -2*t / (1 + t**2)
y = (t**2 - 1) / (1 + t**2)

ratio = sp.simplify(x / (y - 1))
print("x/(y - 1) =", ratio)
print("Verified:", sp.simplify(ratio - t) == 0)
```

#### 3b

```python
# Derivatives and tangent lines
dx_dt = sp.diff(x, t)
dy_dt = sp.diff(y, t)

display("dx/dt =", dx_dt)
display("dy/dt =", dy_dt)

# Horizontal tangent: dy/dt = theta
horizontal_t = sp.solve(dy_dt, t)
print("\nHorizontal tangent at t =", horizontal_t)
print("Points:", [(sp.simplify(x.subs(t, val)), sp.simplify(y.subs(t, val))) for val in horizontal_t])

# Vertical tangent: dx/dt = theta
vertical_t = sp.solve(dx_dt, t)
print("\nVertical tangent at t =", vertical_t)
print("Points:", [(sp.simplify(x.subs(t, val)), sp.simplify(y.subs(t, val))) for val in vertical_t])
```

#### 3c

```python
# Simplify x(t)^2 + y(t)^2
result = sp.simplify(x**2 + y**2)
print("x(t)^2 + y(t)^2 =", result)
```

```python
# Plot for t in [-10, 10] and [-50, 50]
plot_parametric(x, y, (t, -10, 10), title='t in [-10, 10]')
plot_parametric(x, y, (t, -50, 50), title='t in [-50, 50]')
```

####3f

```python
# Limits as t → ∞
limit_x = sp.limit(x, t, sp.oo)
limit_y = sp.limit(y, t, sp.oo)

print("lim(t→∞) x(t) =", limit_x)
print("lim(t→∞) y(t) =", limit_y)
print(f"\nlim(t→∞) f(t) = ({limit_x}, {limit_y}) = P")
print("Define f(∞) = (0, 1) to complete the circle")
```
